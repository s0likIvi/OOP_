# Третє завдання 
## Нfписати програму за допомогою Copilot, запит: 
напиши першу найпростішу програму але не сильно примітивну для початківця. хай це буде щось повʼязане із навчанням, може якийсь список успішності або щось подібне

```python
print("=== Список успішності ===")
student_name = input("Введіть ім'я учня: ")

grades_by_subject = {
    "Математика": [10, 11, 9],
    "Українська мова": [12, 10, 11],
    "Інформатика": [11, 12, 12],
}

all_grades_sum = 0
all_grades_count = 0
best_subject = ""
best_average = 0

print(f"\nОцінки учня: {student_name}")

for subject, grades in grades_by_subject.items():
    average = sum(grades) / len(grades)
    print(f"{subject}: {grades} — середній бал {average:.1f}")

    all_grades_sum += sum(grades)
    all_grades_count += len(grades)

    if average > best_average:
        best_average = average
        best_subject = subject

overall_average = all_grades_sum / all_grades_count

print(f"\nЗагальний середній бал: {overall_average:.1f}")
print(f"Найкращий предмет: {best_subject}")

if overall_average >= 10:
    print("Чудовий результат! Так тримати!")
elif overall_average >= 7:
    print("Добрий результат! Є куди прагнути.")
else:
    print("Варто приділити навчанню трохи більше уваги.")
```
