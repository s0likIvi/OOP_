## 1. Основні типи даних у Python
у Python існують різні типи даних, які використовуються для зберігання інформації. Розглянемо рядки, цілі та дробові числа, списки, кортежі, множини та словники.
Основними типами є:
- int — цілі числа;
- float — дробові числа;
- complex — комплексні числа;
- str — текстові рядки;
- list — списки;
- tuple — кортежі;
- set — множини;
- dict — словники;
- bool — логічні значення.

Тип даних визначає, які операції можна виконувати зі значеннями

```python
# Рядки (str)
a = "Змінна з текстом"
a1 = 'a'

#Числові змінні 
b=1
b1=1.1
complex_number = 2+3j

#Список (list)
c=["a", 1, 1.25, "Слово", a]

#Словник (dict)
d={"a": "Слово", "b": 1, a: b}

#Кортедж (tuple)
e=("a", a)

# Множина (set)
f = {"ss", a + str(b)}

# Логічне значення
g = True

# Значення None
h = None

print("Рядок:", a)
print("Символ:", a1)
print("Ціле число:", b)
print("Дробове число:", b1)
print("Комплексне число:", complex_number)
print("Список:", c)
print("Словник:", d)
print("Кортеж:", e)
print("Множина:", f)
print("Логічне значення:", g)
print("Порожнє значення:", h)

print("\nТипи даних:")

print(type(a))
print(type(b))
print(type(b1))
print(type(complex_number))
print(type(c))
print(type(d))
print(type(e))
print(type(f))
print(type(g))
print(type(h))
```

## 1.1 операції з числами та рядками 
у python можна виконувати арифметичні операції над числами, а також обʼєднувати, повторювати та отриимувати окремі символи рядків.

```python
x = 17
y = 3

print("Додавання:", x + y)
print("Віднімання:", x - y)
print("Множення:", x * y)
print("Ділення:", x / y)
print("Цілочисельне ділення:", x // y)
print("Остача від ділення:", x % y)
print("Піднесення до степеня:", x ** y)

word = "Python"

print("\nОб'єднання рядків:", word + " Basics")
print("Повторення рядка:", word * 2)
print("Перший символ:", word[0])
print("Частина рядка:", word[1:4])
```

## 2. Вбудовані константи та зарезервовані слова

Вбудовані константи - це спеціальні значення, які вже існують у Python.

Зарезервовані слова - це слова, які мають спеціальне призначення в мові програмування. Їх не можна використовувати як назви змінних.

```python
import keyword

print("Перша константа:", True)
print("Друга константа:", False)
print("Третя константа:", None)

print(f"\nРезультат перевірки: {True}")

print("\nЗарезервовані слова Python:")
print(keyword.kwlist)

print("\nПеревірка зарезервованих слів:")
print("for:", keyword.iskeyword("for"))
print("student:", keyword.iskeyword("student"))
```

## 3. Вбудовані функції Python

Вбудовані функції - це готові інструменти Python, які дозволяють виконувати різні операції без створення власних функцій.

```python
# Функція abs() повертає абсолютне значення числа
print("Абсолютне значення:", abs(-12.5))

# Функція len() визначає кількість елементів
fruits = ["яблуко", "банан", "апельсин"]
print("Кількість фруктів:", len(fruits))

# Функція round() округлює число
number = 15.6789
print("Округлене число:", round(number, 2))
```

## 4. Цикли в Python

Цикли використовуються для багаторазового виконання певних команд.

У Python існують два основні цикли:
- for - використовується для перебору елементів послідовності;
- while - виконується, поки задана умова залишається істинною.

```python
# Приклад 1. Цикл for
print("Числа від 1 до 5:")

for i in range(1, 6):
    print(i)

# Приклад 2. Перебирання списку
products = ["хліб", "молоко", "сир", "яблука"]

print("\nСписок покупок:")

for product in products:
    print("Потрібно купити:", product)

# Приклад 3. Цикл while
print("\nЗворотний відлік:")

count = 5

while count > 0:
    print(count)
    count -= 1

print("Старт!")
```

## 5. Розгалуження в Python

Розгалуження дозволяють програмі приймати рішення залежно від заданих умов.

Для створення розгалужень використовуються конструкції if, elif та else.

```python
# Приклад 1. Перевірка віку
age = 17

if age >= 18:
    print("Ви повнолітні.")
else:
    print("Ви ще неповнолітні.")

# Приклад 2. Перевірка оцінки
grade = 85

if grade >= 90:
    print("Відмінно!")
elif grade >= 75:
    print("Добре!")
elif grade >= 60:
    print("Задовільно.")
else:
    print("Потрібно повторити матеріал.")

# Приклад 3. Перевірка парності числа
number = 12

if number % 2 == 0:
    print(f"Число {number} є парним.")
else:
    print(f"Число {number} є непарним.")
```

## 6. Обробка помилок try / except / finally

У Python помилки можуть виникати безпосередньо під час виконання програми.

Конструкція try-except-finally дозволяє обробляти такі ситуації:
- try - містить код, у якому може виникнути помилка;
- except - перехоплює та обробляє помилку;
- finally - виконується незалежно від того, чи виникла помилка.

```python
number = 0

try:
    result = 100 / number
    print("Результат:", result)

except ZeroDivisionError as error:
    print("Виникла помилка:", error)

finally:
    print("Перевірку завершено.")
```

## 7. Контекстний менеджер with

Контекстний менеджер with використовується для правильного керування ресурсами.

Наприклад, під час роботи з файлами він автоматично закриває файл після завершення роботи з ним.

```python
# Створення файлу та запис інформації
with open("example.txt", "w", encoding="utf-8") as file:
    file.write("Я вивчаю мову програмування Python.\n")
    file.write("Контекстний менеджер with допомагає працювати з файлами.")

# Читання інформації з файлу
with open("example.txt", "r", encoding="utf-8") as file:
    for number, line in enumerate(file, start=1):
        print(f"{number}. {line.strip()}")
```

## 8. Лямбда-функції Python

Лямбда-функції - це невеликі анонімні функції, які створюються за допомогою ключового слова lambda.

Вони можуть приймати аргументи та повертати результат одного виразу. Лямбда-функції зручні для коротких операцій.

```python
# Звичайна функція
def calculate_sum(a, b):
    return a + b

# Лямбда-функція
multiply = lambda a, b: a * b

# Ще одна лямбда-функція
square = lambda number: number ** 2

print("Результат звичайної функції:", calculate_sum(5, 7))
print("Результат лямбда-функції:", multiply(5, 7))
print("Квадрат числа 6:", square(6))

# Використання lambda для сортування
students = [("Олена", 90), ("Іван", 75), ("Марія", 96)]

sorted_students = sorted(
    students,
    key=lambda student: student[1],
    reverse=True
)

print("Результат сортування:", sorted_students)
```
